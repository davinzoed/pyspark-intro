# Data Writing with Pyspark

```python
from pyspark.sql.types import *
from pyspark.sql.functions import *
```

```python
df = (
    spark.read
    .format("csv")
    .option("header", True)
    .option("inferSchema", True)
    .load("/Volumes/workspace/introduction/raw-data/BigMart Sales.csv")
)
```

```python
display(df.limit(11))
```

## DATA WRITING

```python
"""
Syntax Dasar
df.write.format("format") \
    .mode("mode") \
    .save("path")

| Parameter  | Fungsi                                 |
| ---------- | -------------------------------------- |
| `format()` | Format file yang akan ditulis          |
| `mode()`   | Apa yang dilakukan jika data sudah ada |
| `save()`   | Lokasi penyimpanan                     |

| Format  | Kapan Dipakai                                 |
| ------- | --------------------------------------------- |
| csv     | File CSV                                      |
| parquet | Format kolumnar yang cepat                    |
| delta   | Format Delta Lake (paling umum di Databricks) |
| json    | JSON                                          |
| orc     | ORC                                           |

Ada 4 mode:
append, overwrite, error (default), ignore

Jika df nya ada headernya maka:
df.write.format("format") \
    .option("header", True) \
    .mode("mode") \
    .save("path")

"""
```

```python
df = df.withColumn("Wuthering_Waves", lit("Suisui Menang 50/50 Aamiinn!"))
df = df.withColumn("Current_Date", current_date())
```

### Append

```python
df.write.format("csv") \
        .mode("append") \
        .option("Header", True) \
        .save("/Volumes/workspace/introduction/output/csv")
```

### Overwrite

```python
df.write.format("csv") \
        .mode("overwrite") \
        .option("Header", True) \
        .save("/Volumes/workspace/introduction/output/spark_intro")
```

### Error

```python
df.write.format("csv") \
        .option("Header", True) \
        .save("/Volumes/workspace/introduction/output/spark_intro")
```

### Parquet

```python
df.write.format("parquet") \
        .mode("error") \
        .option("Header", True) \
        .save("/Volumes/workspace/introduction/output/parquet")
```

### Table

```python
df.write.format("delta") \
        .mode("overwrite") \
        .saveAsTable("workspace.introduction.df_table")
```

## SPARK SQL

```python
"""
Spark SQL adalah modul di Apache Spark yang memungkinkan Anda mengolah data menggunakan sintaks SQL, bukan hanya API PySpark.

Bayangkan ada dua cara untuk mengolah data yang sama.
Menggunakan PySpark DataFrame API
    df.filter(col("amount") > 1000)

Menggunakan SQL
    SELECT *
    FROM sales
    WHERE amount > 1000

Hasilnya sama.
Perbedaannya hanya pada bahasa yang digunakan.


=============== Langkah Menggunakan Spark SQL =============== 
Ada 3 langkah utama:
- Langkah 1: Punya DataFrame
- Langkah 2: Register DataFrame sebagai Temporary View
        df.createOrReplaceTempView("bigmart")
  Sekarang Spark mengenal view bernama: bigmart
- Langkah 3: Jalankan SQL
  Menggunakan spark.sql() jika ingin menyimpan hasil query ke DataFrame


======================= Mindset ======================= 
Anggap Spark SQL sebagai bahasa lain untuk berbicara dengan Spark.

          Spark
             ▲
      ┌──────┴──────┐
      │             │
 PySpark API    Spark SQL

Anda bebas memilih sesuai kebutuhan. Bahkan dalam satu notebook Databricks, keduanya sering dipakai bersama:
    Baca data dengan PySpark.
    Buat temporary view.
    Jalankan query SQL yang kompleks.
    Hasil query kembali menjadi DataFrame PySpark untuk diproses lebih lanjut.

Ringkasan
- Spark SQL memungkinkan Anda mengolah DataFrame menggunakan sintaks SQL.
- Gunakan createOrReplaceTempView() untuk mendaftarkan DataFrame sebagai view sementara.
- Gunakan spark.sql() untuk menjalankan query SQL dari Python.
- Hasil spark.sql() adalah DataFrame PySpark.
- Temporary view bersifat sementara, sedangkan table yang dibuat dengan saveAsTable() bersifat permanen.
- Spark SQL dan DataFrame API umumnya memiliki performa yang setara karena keduanya menggunakan engine optimasi Spark yang sama.
"""
```

```python
df.createOrReplaceTempView("my_view")
```

```python
%sql
SELECT
    Item_Identifier,
    Wuthering_Waves,
    Current_Date
FROM my_view
```

```python
df_sql = spark.sql(
"""
    SELECT
        Item_Identifier,
        Wuthering_Waves,
        Current_Date
    FROM my_view
    LIMIT 70
""")
```
